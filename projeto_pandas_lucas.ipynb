```python
# Para identificar os bairros com maior oferta de imóveis acessíveis, mostre a quantidade de imóveis cadastrados em cada bairro, 
# considerando apenas aqueles com valor abaixo de 250000. Exiba somente os bairros que possuem pelo menos 3 imóveis.

df_imoveis_por_bairros = df[df["valor"] < 250000]

df_imoveis_por_bairros = df_imoveis_por_bairros.groupby("bairro").size()

df_imoveis_por_bairros = df_imoveis_por_bairros[df_imoveis_por_bairros >=3]

df_imoveis_por_bairros
```

```python
# A imobiliária quer saber quais bairros de Belo Horizonte possuem maior volume financeiro em imóveis residenciais. 
# Calcule o valor total dos imóveis de cada bairro, considerando apenas casas, apartamentos e sobrados adquiridos após 2021. (Ordenados do maior valor para o menor)
# repare que aqui eu tenho que usar no filtro  o & e o | que é o or . Tenho que por também entre 2 parentêses para não bagunçara lógica. 

df_bh_bairros_maiores_valores = df[(df["cidade"]== "Belo Horizonte") & ((df["tipo"] == "Casa") | (df["tipo"] == "Apartamento") | (df["tipo"] == "Sobrado")) & (df["data_aquisicao"] > "2021-12-31")]

df_bh_bairros_maiores_valores = df_bh_bairros_maiores_valores.groupby("bairro")["valor"].sum().sort_values(ascending = False)

df_bh_bairros_maiores_valores 
 
```

```python
# Exiba um gráfico de pizza (pie) do valor total dos imóveis por tipo.
# não tem ()  no  plot porque o plot é um acesso a biblioteca matplotlib , o que é a função é o pie.

df_total_tipo = df.groupby("tipo")["valor"].sum()

df_total_tipo.plot.pie()
```

```python
# Exiba um gráfico de barras horizontal (barh) do valor total dos imóveis de cada  bairro da cidade "Sao Paulo".
# note que a soma  sempre vem depois do filtro conforme abaixo. 
#  aqui eu não posso fazer assim df_sp = df[(df["cidade"] == "Sao Paulo") & (df.groupby("bairro")["valor"].sum())]  porque o & só junta dois filtros 
# e não  um f iltro com o groupby.  Se eu quiser  fazer em uma linha só, ficaria assim : 
# df_sp = df[(df["cidade"] == "Sao Paulo")].groupby("bairro")["valor"].sum().sort_values(ascending=False)


df_sp = df[(df["cidade"] == "Sao Paulo")]

df_sp = df_sp.groupby("bairro")["valor"].sum().sort_values(ascending=False)

df_sp.plot.barh()
```

```python
# Crie as colunas ano e mes. Utilize df.data_aquisicao.dt.year e df.data_aquisicao.dt.month.
# vamos lá , a primeira informação importante aqui é que o dt é um comando que extrai partes de uma data de uma coluna no DF . Ele é como o  astype(str).str.startswith
#  mas ao invés de texto pega a data .  
# outra informação aqui é que o enunciado pede para criar colunas no DF,  logo  a variável que eu crio tem que ser uma variável com o nome das  coluans que quero criar
# entre []  e com ""  conforme abaixo . 
#  esse exericio até da pra fazer sem variável também  mas  não faz sentido  porque provavelmente vou querer guardar essas duas colunas que coloquei . 
# uma coisa que achei estranho é que aqui  eu coloquei duas colunas no DF original ,  não teria que ser um novo DF ? 
# aqui o ano é um número assim como  o mês. 

df["ano"] = df["data_aquisicao"].dt.year
df["mes"]= df["data_aquisicao"].dt.month

df
```

```python
# Exiba um gráfico de linhas com a evolução do valor total dos imóveis agrupados por ano.
#  veja que como o exercicio pede o gráfico por linha ,  eu  vou usar a função line() e de novo com o acessador plot. 
# aqui eu não ordenei porque ele já vem ordenado pelo ano  por conta do groupby , o groupby sempre ordena pelo índice.

df_valor_ano = df.groupby("ano")["valor"].sum()

df_valor_ano.plot.line()
```

```python
# Exiba um gráfico de barras com a quantidade de imóveis adquiridos em cada ano.

df_qtd_imóveis_ano = df.groupby("ano").size()

df_qtd_imóveis_ano.plot.bar()
```

```python
#Calcule o valor total dos imóveis por ano, considerando apenas casas, apartamentos e sobrados adquiridos. 
# Exiba um gráfico de barras agrupado por ano e tipo de imóvel.
# note que aqui não precisei usar o sort_values  ano  porque no groupby  ele já meio que ordena por ano também .
# note 

total_imoveis_ano = df[(df["tipo"] == "Casa") | (df["tipo"] == "Apartamento") | (df["tipo"] == "Sobrado")]

total_imoveis_ano = total_imoveis_ano.groupby(["ano","tipo"])["valor"].sum().unstack()

total_imoveis_ano

total_imoveis_ano.plot.bar()
```
