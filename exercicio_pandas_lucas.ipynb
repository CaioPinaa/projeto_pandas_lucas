```python
# Exiba a média de valores dos imóveis da cidade "Belo Horizonte".
# Exiba a média de valores dos imóveis da cidade "Belo Horizonte".
# Aqui eu criei sem nenhuma variável 
# aqui eu não separei por & porque só tem um filtro 
# o que caracteriza um filtro é ter uma comparação (==, >=, <=, >, <, !=)

df[df["cidade"]=="Belo Horizonte"]["valor"].mean()
```

```python
# Exiba os imóveis com valor entre 300000 e 600000 da cidade Curitiba.
# sem variável 
# Conforme aprendi  o () chama a função e o [] acessa uma coluna e filtra dados também . Só que aqui  achamos mais uma função para () que seria separar as condições e isso é comum
# quando tem o & no script. 
# aqui eu  usei   & porque tem mais de um filtro.
# # o que caracteriza um filtro é ter uma comparação (==, >=, <=, >, <, !=) 

df[(df["cidade"] == "Curitiba") & (df["valor"] >= 300000) & (df["valor"] <= 600000)]
```

```python
# Exiba os imóveis com valor entre 300000 e 600000 da cidade Curitiba.
# aqui eu consegui fazer criando uma variável 

df_curitiba = df[df["cidade"] == "Curitiba"]

df_curitiba = df_curitiba[(df_curitiba["valor"] >= 300000) & (df_curitiba["valor"] <= 600000)]

df_curitiba
```
